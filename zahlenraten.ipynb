```python
import random  

geheime_zahl = random.randint(1, 10) 

geratene_zahl = 0 

print("Willkommen beim Zahlenraten mit Zufall!")
print("Der Computer hat sich eine Zahl zwischen 1 und 10 ausgedacht.")

while geratene_zahl != geheime_zahl:
    eingabe = input("Dein Tipp: ")
    geratene_zahl = int(eingabe)
    
    if geratene_zahl < geheime_zahl:
        print("Zu niedrig! Versuchs nochmal.")
    elif geratene_zahl > geheime_zahl:
        print("Zu hoch! Versuchs nochmal.")
    else:
        print("Richtig! Du hast die Zahl erraten!")

print("Spiel beendet.")
```
